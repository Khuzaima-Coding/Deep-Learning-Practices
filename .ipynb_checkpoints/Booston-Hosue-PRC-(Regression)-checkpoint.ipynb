## Loading Boaston Housing Dataset

```python
import pandas as pd
import numpy as np
from tensorflow import keras 
from keras import layers
```

```python
from tensorflow.keras.datasets import boston_housing
(train_data, train_targets), (test_data, test_targets) = (boston_housing.load_data())
```

## Normlaization(standardization) of Data

```python
mean = train_data.mean(axis=0)
train_data-= mean
std = train_data.std(axis=0)
train_data /= std
test_data-= mean
test_data /= std
#These things are done by standard scaler or minmaxscaler functions
```

## Model Defining

```python
def build_model():
    model = keras.Sequential([   
    layers.Dense(64, activation="relu"),
    layers.Dense(64, activation="relu"),
    layers.Dense(1)
    #Because we need to instantiate the same model multiple times,we use a function to construct it.
        ])
    model.compile(optimizer="rmsprop", loss="mse", metrics=["mae"])
    return model
```

```python
k = 4

num_val_samples = len(train_data) // k

num_epochs = 100

all_scores = []

for i in range(k):
    print(f"Processing fold #{i}")

    # Validation data
    val_data = train_data[
        i * num_val_samples : (i + 1) * num_val_samples
    ]

    val_targets = train_targets[
        i * num_val_samples : (i + 1) * num_val_samples
    ]

    # Training data
    partial_train_data = np.concatenate(
        [
            train_data[:i * num_val_samples],
            train_data[(i + 1) * num_val_samples:]
        ],
        axis=0
    )

    partial_train_targets = np.concatenate(
        [
            train_targets[:i * num_val_samples],
            train_targets[(i + 1) * num_val_samples:]
        ],
        axis=0
    )
     # Build a new model
    model = build_model()
    model.fit(

    # Train
        partial_train_data,
        partial_train_targets,
        epochs=num_epochs,
        batch_size=16,
        verbose=0
     )

    # Evaluate
    val_mse, val_mae = model.evaluate(
        val_data,
        val_targets,
        verbose=0
     )

    # Store score
    all_scores.append(val_mae)
```
