```python
from ultralytics import YOLO   # you only look once
```

```python
model = YOLO("yolo11n.pt")
```

```python
pred= model.predict("/content/lapto.jpg")
```

```python
pred[0].plot()
```

```python
import gradio as gr
```

```python
def prediction_image(image):
  prediction  = model.predict(image)
  return prediction[0].plot()
```

```python
app = gr.Interface(fn =prediction_image, inputs="image", outputs="image", title = "YOLO MODEL PREDICTION" )
```

```python
app.launch(inbrowser=True, inline=False)
```
