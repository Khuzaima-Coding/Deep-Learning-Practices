```python
#Multi Hot Encode Our Data
import numpy as np 
def vectorize_sequences(sequences, dimension=10000):   #It selects only 10000 words from vocablary,instead of using 88584
    results = np.zeros((len(sequences), dimension))
    for i, sequence in enumerate(sequences):
        for j in sequence:
            results[i, j] = 1.0           
    return results
```

```python
x_train = vectorize_sequences(train_data)   
x_test = vectorize_sequences(test_data)
```

```python
from tensorflow import keras 
from keras import layers
model=keras.Sequential([
    layers.Dense(16,activation="relu"),
    layers.Dense(16,activation="relu"),
    layers.Dense(1,activation="sigmoid")
])
```

## Plot For Training and Validation Loss

```python
import matplotlib.pyplot as plt
history_dict = history.history
loss_values = history_dict["loss"]
val_loss_values = history_dict["val_loss"]
epochs = range(1, len(loss_values) + 1)
plt.plot(epochs, loss_values, "bo", label="Training loss")
plt.plot(epochs, val_loss_values, "b", label="Validation loss")   
plt.title("Training and validation loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
```

## Plot For Training and Validation Accuracy

```python
history_dict = history.history
loss_values = history_dict["accuracy"]
val_loss_values = history_dict["val_accuracy"]
epochs = range(1, len(loss_values) + 1)
plt.plot(epochs, loss_values, "bo", label="Training Accuracy")  #Here "bo" means b and circular dots 
plt.plot(epochs, val_loss_values, "b", label="Validation Accuracy")   #Here "b" means blue line
plt.title("Training and validation Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.show()
```

# Lets we can make pipeline of This
So a user can add the directly text and it gives the output

```python
def load_imdb_vocab():
    word_index=imdb.get_word_index()
    word_index={k:(v+3) for k,v in word_index.items()} #The dictionary comprehension is like as the list comprehension.
    word_index["<PAD>"]=0
    word_index["<START>"]=1
    word_index["<UNK>"]=2
    word_index["<UNUSED>"]=3
    return word_index
```

```python
def text_to_sequence(text,word_index,max_words=10000):
    words=text.lower().split()
    encoded=[1] #<START>

    for w in words:
        idx=word_index.get(w,2) #<UNK>
        encoded.append(idx)

    return encoded
```

```python
# text="I am khuzaima"
# text.lower().split()
```

```python
import numpy as np

def vectorize(sequence,dimension=10000):
    result=np.zeros((1,dimension))
    for index in sequence:
        if index < dimension:
            result[0,index]=1.0
    return result
```

```python
def imdb_sentiment_pipeline(text,model,word_index,max_words=10000):
    seq=text_to_sequence(text,word_index,max_words=max_words)
    vec=vectorize(seq,dimension=max_words)
    pred=model.predict(vec)[0][0]  #it says that takes [[number]] 0 index array and  gives [num],then 0 mean its 0 index, means num
    sentiment="positive" if pred > 0.5 else "negative"
    return sentiment,float(pred)
```

```python
word_index=load_imdb_vocab()
# text="This movie is great"
text="This movie is not a great movie!"  #Becasue it is bag of word model(not a sequential model),so it gives us positive sentiment on the word basis

sentiment,score=imdb_sentiment_pipeline(text,model,word_index)
print(sentiment,score)
```
