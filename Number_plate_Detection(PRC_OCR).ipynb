Improved Video code

```python
# =========================================================
# 1. IMPORT LIBRARIES
# =========================================================

import cv2
import os
import re
import torch
import easyocr

from ultralytics import YOLO
from collections import Counter, deque


# =========================================================
# 2. CHECK GPU
# =========================================================

USE_GPU = torch.cuda.is_available()

if USE_GPU:
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("GPU not available - using CPU")


# =========================================================
# 3. LOAD YOLO MODEL
# =========================================================

model = YOLO(
    "/content/best.pt"
)

print("YOLO model loaded")


# =========================================================
# 4. LOAD EASY OCR
# =========================================================

reader = easyocr.Reader(
    ['en'],
    gpu=USE_GPU
)

print("EasyOCR loaded")


# =========================================================
# 5. INPUT VIDEO
# =========================================================

input_video = "/content/16477211_1920_1080_25fps.mp4"

cap = cv2.VideoCapture(input_video)

if not cap.isOpened():
    raise Exception("Could not open input video")


# =========================================================
# 6. GET VIDEO INFORMATION
# =========================================================

fps = cap.get(cv2.CAP_PROP_FPS)

width = int(
    cap.get(cv2.CAP_PROP_FRAME_WIDTH)
)

height = int(
    cap.get(cv2.CAP_PROP_FRAME_HEIGHT)
)

total_frames = int(
    cap.get(cv2.CAP_PROP_FRAME_COUNT)
)


print()
print("====================================")
print("VIDEO INFORMATION")
print("====================================")
print("FPS:", fps)
print("Width:", width)
print("Height:", height)
print("Total frames:", total_frames)
print("====================================")


# =========================================================
# 7. CREATE TEMPORARY AVI OUTPUT
# =========================================================

temp_output = "/content/temp_license_plate.avi"

fourcc = cv2.VideoWriter_fourcc(*"MJPG")

out = cv2.VideoWriter(
    temp_output,
    fourcc,
    fps,
    (width, height)
)

if not out.isOpened():
    raise Exception("VideoWriter could not be opened")


# =========================================================
# 8. OCR SETTINGS
# =========================================================

# License plate characters
allowlist = (
    "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    "0123456789"
)

# OCR confidence threshold
OCR_CONFIDENCE = 0.35

# Run OCR every N frames for an existing track
OCR_INTERVAL = 5

# Number of OCR predictions stored for each plate
HISTORY_SIZE = 7

# Remove a track after it has disappeared for this many frames
MAX_MISSED_FRAMES = 30


# =========================================================
# 9. STORE INFORMATION SEPARATELY FOR EACH PLATE
# =========================================================

plate_states = {}


# =========================================================
# Each plate gets its own dictionary:
#
# plate_states[track_id] = {
#     "history": deque(...),
#     "last_text": "",
#     "last_confidence": 0,
#     "last_ocr_frame": -9999,
#     "last_seen": frame_count
# }
# =========================================================


# =========================================================
# 10. OCR TEXT CLEANING FUNCTION
# =========================================================

def clean_text(text):

    text = text.upper()

    # Remove spaces
    text = text.replace(" ", "")

    # Keep only A-Z and 0-9
    text = re.sub(
        r"[^A-Z0-9]",
        "",
        text
    )

    return text


# =========================================================
# 11. OCR FUNCTION
# =========================================================

def read_plate(plate):

    if plate is None or plate.size == 0:
        return None, 0.0


    # -----------------------------------------------------
    # Enlarge plate
    # -----------------------------------------------------

    enlarged = cv2.resize(
        plate,
        None,
        fx=4,
        fy=4,
        interpolation=cv2.INTER_CUBIC
    )


    # -----------------------------------------------------
    # Convert to grayscale
    # -----------------------------------------------------

    gray = cv2.cvtColor(
        enlarged,
        cv2.COLOR_BGR2GRAY
    )


    # -----------------------------------------------------
    # Improve contrast
    # -----------------------------------------------------

    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )

    enhanced = clahe.apply(gray)


    # -----------------------------------------------------
    # Only TWO versions
    #
    # This is intentional.
    # We do not run 4 versions + rotations because
    # that would unnecessarily increase computation.
    # -----------------------------------------------------

    versions = [
        enlarged,
        enhanced
    ]


    candidates = []


    # =====================================================
    # Run OCR
    # =====================================================

    for image in versions:

        results = reader.readtext(
            image,
            detail=1,
            paragraph=False,
            allowlist=allowlist
        )


        # -------------------------------------------------
        # Process OCR results
        # -------------------------------------------------

        for detection in results:

            text = detection[1]
            confidence = float(detection[2])


            # Clean text
            text = clean_text(text)


            # Ignore useless results
            if len(text) < 2:
                continue


            # Ignore low confidence results
            if confidence < OCR_CONFIDENCE:
                continue


            candidates.append(
                (
                    text,
                    confidence
                )
            )


    # =====================================================
    # No valid OCR result
    # =====================================================

    if len(candidates) == 0:
        return None, 0.0


    # =====================================================
    # Select highest-confidence OCR result
    # =====================================================

    best_text, best_confidence = max(
        candidates,
        key=lambda x: x[1]
    )


    return best_text, best_confidence


# =========================================================
# 12. VIDEO PROCESSING
# =========================================================

frame_count = 0


while True:

    # -----------------------------------------------------
    # Read frame
    # -----------------------------------------------------

    ret, frame = cap.read()

    if not ret:
        break


    frame_count += 1


    # =====================================================
    # YOLO TRACKING
    # =====================================================
    #
    # ByteTrack assigns an ID to each detected plate.
    #
    # Example:
    #
    # Plate A -> ID 1
    # Plate B -> ID 2
    #
    # Their OCR histories remain separate.
    # =====================================================

    results = model.track(
        frame,
        persist=True,
        tracker="bytetrack.yaml",
        verbose=False,
        device=0 if USE_GPU else "cpu"
    )


    current_track_ids = set()


    # =====================================================
    # PROCESS YOLO RESULT
    # =====================================================

    for result in results:

        boxes = result.boxes


        # -------------------------------------------------
        # Check whether plates were detected
        # -------------------------------------------------

        if boxes is None or len(boxes) == 0:
            continue


        # -------------------------------------------------
        # Get bounding boxes
        # -------------------------------------------------

        xyxy = boxes.xyxy


        # -------------------------------------------------
        # Get tracking IDs
        # -------------------------------------------------

        track_ids = boxes.id


        # =================================================
        # Process each detected plate
        # =================================================

        for i, box in enumerate(xyxy):


            # -------------------------------------------------
            # Get coordinates
            # -------------------------------------------------

            x1, y1, x2, y2 = map(
                int,
                box
            )


            # -------------------------------------------------
            # Keep coordinates inside frame
            # -------------------------------------------------

            x1 = max(0, x1)
            y1 = max(0, y1)

            x2 = min(width, x2)
            y2 = min(height, y2)


            # -------------------------------------------------
            # Draw detection box
            # -------------------------------------------------

            cv2.rectangle(
                frame,
                (x1, y1),
                (x2, y2),
                (0, 255, 0),
                3
            )


            # -------------------------------------------------
            # Get tracking ID
            # -------------------------------------------------

            if track_ids is not None:

                track_id = int(
                    track_ids[i].item()
                )

                current_track_ids.add(
                    track_id
                )

            else:

                track_id = None


            # =================================================
            # CROP PLATE FROM ORIGINAL FRAME
            # =================================================

            plate = frame[
                y1:y2,
                x1:x2
            ]


            if plate.size == 0:
                continue


            # =================================================
            # CREATE STATE FOR NEW TRACK
            # =================================================

            if track_id is not None:

                if track_id not in plate_states:

                    plate_states[track_id] = {

                        "history": deque(
                            maxlen=HISTORY_SIZE
                        ),

                        "last_text": "",

                        "last_confidence": 0.0,

                        "last_ocr_frame": -9999,

                        "last_seen": frame_count
                    }


                # Update last seen frame
                plate_states[track_id][
                    "last_seen"
                ] = frame_count


                state = plate_states[
                    track_id
                ]


                # =================================================
                # DECIDE WHETHER OCR SHOULD RUN
                # =================================================
                #
                # New plate:
                #     OCR immediately
                #
                # Existing plate:
                #     OCR every 5 frames
                # =================================================

                should_run_ocr = (

                    frame_count
                    -
                    state["last_ocr_frame"]

                    >= OCR_INTERVAL
                )


                if should_run_ocr:

                    # -------------------------------------------------
                    # Run OCR
                    # -------------------------------------------------

                    text, confidence = read_plate(
                        plate
                    )


                    # Remember when OCR was performed
                    state["last_ocr_frame"] = (
                        frame_count
                    )


                    # =================================================
                    # Valid OCR result
                    # =================================================

                    if text is not None:

                        # ---------------------------------------------
                        # Add result to THIS plate's history
                        # ---------------------------------------------

                        state["history"].append(
                            text
                        )


                        # ---------------------------------------------
                        # Find most common prediction
                        # ---------------------------------------------

                        most_common_text = (
                            Counter(
                                state["history"]
                            )
                            .most_common(1)[0][0]
                        )


                        # ---------------------------------------------
                        # Save stable result
                        # ---------------------------------------------

                        state["last_text"] = (
                            most_common_text
                        )

                        state["last_confidence"] = (
                            confidence
                        )


                        # ---------------------------------------------
                        # Print OCR result
                        # ---------------------------------------------

                        print(
                            f"Frame {frame_count} | "
                            f"Plate ID: {track_id} | "
                            f"Text: {state['last_text']} | "
                            f"Confidence: "
                            f"{confidence:.2f}"
                        )


            # =================================================
            # DISPLAY OCR RESULT
            # =================================================

            if track_id is not None:

                state = plate_states[
                    track_id
                ]


                display_text = ""


                if state["last_text"] != "":

                    display_text = (
                        state["last_text"]
                        +
                        " ("
                        +
                        str(
                            round(
                                state[
                                    "last_confidence"
                                ],
                                2
                            )
                        )
                        +
                        ")"
                    )


                # -------------------------------------------------
                # Display OCR text above plate
                # -------------------------------------------------

                if display_text != "":

                    cv2.putText(
                        frame,
                        display_text,
                        (
                            x1,
                            max(
                                y1 - 15,
                                30
                            )
                        ),
                        cv2.FONT_HERSHEY_SIMPLEX,
                        0.9,
                        (0, 255, 0),
                        3
                    )


                # -------------------------------------------------
                # Display tracking ID
                # -------------------------------------------------

                # cv2.putText(
                #     frame,
                #     f"ID: {track_id}",
                #     (
                #         x1,
                #         min(
                #             y2 + 30,
                #             height - 10
                #         )
                #     ),
                #     cv2.FONT_HERSHEY_SIMPLEX,
                #     0.7,
                #     (255, 255, 0),
                #     2
                # )


    # =====================================================
    # REMOVE OLD TRACKS
    # =====================================================

    tracks_to_delete = []


    for track_id, state in plate_states.items():

        if (
            frame_count
            -
            state["last_seen"]
            >
            MAX_MISSED_FRAMES
        ):

            tracks_to_delete.append(
                track_id
            )


    for track_id in tracks_to_delete:

        del plate_states[
            track_id
        ]


    # =====================================================
    # SAVE FRAME
    # =====================================================

    out.write(frame)


    # =====================================================
    # PROGRESS
    # =====================================================

    if frame_count % 100 == 0:

        print(
            f"Processed "
            f"{frame_count}/"
            f"{total_frames} frames"
        )


# =========================================================
# 13. RELEASE VIDEO
# =========================================================

cap.release()
out.release()


# =========================================================
# 14. CHECK OUTPUT
# =========================================================

print()
print("====================================")
print("PROCESSING COMPLETED")
print("====================================")
print("Frames processed:", frame_count)
print("Temporary output:", temp_output)

if os.path.exists(temp_output):

    print(
        "File size:",
        os.path.getsize(temp_output),
        "bytes"
    )

print("====================================")
```

```python
import cv2
import easyocr
import numpy as np
from ultralytics import YOLO
from matplotlib import pyplot as plt

# Load YOLO model
model = YOLO("/content/runs/detect/train/weights/best.pt")

# Load OCR
reader = easyocr.Reader(['en'])


# -----------------------------
# Load image
# -----------------------------

image_path = "/content/OIP (1).webp"

image = cv2.imread(image_path)

if image is None:
    raise Exception("Could not open image")


# Keep original image for display
output = image.copy()


# -----------------------------
# YOLO detection
# -----------------------------

results = model(image, verbose=False)


# -----------------------------
# Process detected plates
# -----------------------------

for result in results:

    for box in result.boxes.xyxy:

        # Get coordinates
        x1, y1, x2, y2 = map(int, box)

        # Keep coordinates inside image
        x1 = max(0, x1)
        y1 = max(0, y1)
        x2 = min(image.shape[1], x2)
        y2 = min(image.shape[0], y2)

        # -----------------------------
        # Draw YOLO bounding box
        # -----------------------------

        cv2.rectangle(
            output,
            (x1, y1),
            (x2, y2),
            (0, 255, 0),
            3
        )

        # -----------------------------
        # Crop license plate
        # -----------------------------

        plate = image[y1:y2, x1:x2]

        if plate.size == 0:
            continue

        # -----------------------------
        # ENLARGE PLATE
        # -----------------------------

        plate = cv2.resize(
            plate,
            None,
            fx=3,
            fy=3,
            interpolation=cv2.INTER_CUBIC
        )

        # -----------------------------
        # Convert to grayscale
        # -----------------------------

        gray = cv2.cvtColor(
            plate,
            cv2.COLOR_BGR2GRAY
        )

        # -----------------------------
        # Improve contrast
        # -----------------------------

        gray = cv2.equalizeHist(gray)

        # -----------------------------
        # OCR
        # -----------------------------

        ocr_result = reader.readtext(
            gray,
            detail=1,
            paragraph=False
        )

        print("\nOCR RESULTS:")

        if len(ocr_result) == 0:

            print("No text detected by OCR")

        else:

            for detection in ocr_result:

                bbox = detection[0]
                text = detection[1]
                confidence = detection[2]

                print(
                    "Text:",
                    text,
                    "| Confidence:",
                    confidence
                )

                # -----------------------------
                # Put OCR text on original image
                # -----------------------------

                cv2.putText(
                    output,
                    text,
                    (x1, max(y1 - 15, 30)),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    1,
                    (0, 255, 0),
                    3
                )


# -----------------------------
# Display final image
# -----------------------------

image_rgb = cv2.cvtColor(
    output,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(12, 8))
plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```
