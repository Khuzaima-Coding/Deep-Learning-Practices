```python
from tensorflow import keras
from tensorflow.keras import layers
model=keras.Sequential([                              #There we can add the layers in list form
    layers.Dense(512,activation="relu"),
    layers.Dense(10,activation="softmax")
])
```

```python
#There we add the layers step by step
from tensorflow.keras import models
from tensorflow.keras import layers
network=models.Sequential()
network.add(layers.Dense(512,activation="relu"))
network.add(layers.Dense(10,activation="softmax"))
```

```python
model.compile(optimizer="rmsprop",loss="categorical_crossentropy",metrics=["accuracy"])
#In this optimizer can do the backpropagation,means in reverse direction it can changes the weights.
# SGD
# RMSprop
# Adam
# Adam
# Adadelta
# Adagrad
# Adamax
# Adafactor
# Nadam
# Ftrl
# These are multiple optimizers ,they have different matematical techniques to optimize weights.

#Loss is the difference between the predicted and original value and categorical_crossentropy is its function for loss.
#metrics=["accuracy"] is the performance matrics.by which we can the performance of our data.
```

```python
train_images = train_images.reshape(60000,784)   # we do this for converting our values into vector form ,bec our Dense layer required this
train_images=train_images.astype("float")/255   #we can divide our images values by 255 ,Bec our maximum value is 255 in rgb colors(in pixels).
test_images=test_images.reshape(10000,784)
test_images=test_images.astype("float")/255     #we do this for coverting the values from 0 till 1
```

```python
model.fit(train_images,train_labels,epochs=5,batch_size=128)
#epochs can do the back propagation ,till 5 times.
#batch_size=128 ,It can select the 128 images(vectors),at each time,and can update the whole batch weight at  once,that can speed up our propagations.
```
