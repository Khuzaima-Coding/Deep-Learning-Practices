## 1️⃣ Component 1: Model
Model AI ka dimaag (brain) hota hai jo hamare sawalat ko samajhta hai aur response generate karta hai.

### 🧠 Gemini vs Groq
* **Gemini (by Google):** Yeh Google ka kafi advanced aur smart model hai jo deep reasoning aur complex tasks ke liye bohot behtareen hai.
* **Groq (Hardware/Hosting Platform):** Groq khud model nahi banata, balki Llama 3 ya Mixtral jaise open-source models ko apne super-fast chips (LPUs) par host karta hai. Iska response time bohot hi fast (miliseconds mein) hota hai.

### 🌡️ Temperature kya hota hai?
Temperature model ki **creativity** aur **randomness** ko control karta hai:
* **Low Temperature (e.g., 0.0):** Model bohot precise, factual, aur predictable answers deta hai (best for coding ya math).
* **High Temperature (e.g., 0.7 - 1.0):** Model zyada creative, unique, aur imaginative ho jata hai (best for story writing ya brainstorming).

### Gemini Model Code Example

```python
# Google Gemini model direct initialization
from langchain_google_genai import ChatGoogleGenerativeAI

gemini_model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key="paste the API There"
)

# Direct call
response = gemini_model.invoke("Ek line mein batao AI kya hai in english?")
print("Gemini Response:", response.content)
```

## 2️⃣ Component 2: Prompt Templates
Prompt woh instruction hoti hai jo hum model ko dete hain. LangChain mein hum prompts ko handle karne ke liye structured templates use karte hain.

### 💬 Message Types
Chat models mein 3 basic messages hote hain:
1. **System Message:** Yeh model ka role decide karta hai (jaise, "Tum ek mathematical genius ho").
2. **Human Message:** User ka actual sawal ya prompt (jaise, "2+2 kya hota hai?").
3. **AI Message:** Model ka response jo AI return karta hai.

### ⚡ Static vs Dynamic Prompts
* **Static Prompt:** Yeh fixed hote hain. Har baar same input aur instruction jati hai (e.g., "Tell me a joke.").
* **Dynamic Prompt:** Yeh templates hote hain jahan hum variables pass karte hain. input ke mutabiq prompt change ho jata hai (e.g., "Tell me a joke about {}").

## 3️⃣ Component 3: Chain
Chain ka matlab hai different components ko ek sath jodna (connect karna).

### 🔗 LCEL and `|` (Pipe) Operator
LangChain Expression Language (LCEL) mein hum `|` (pipe) operator use karte hain.
Iska kaam bilkul assembly line jaisa hai: ek step ka output automatically doosre step ka input ban jata hai.

`chain = prompt | model`

Is line ka matlab hai: Pehle `prompt` dynamic variables ke sath fill hoga, phir jo formatted string banegi woh seedha `model` ko pass ho jayegi aur response generate hoga.

### Simple Chain Example

### 📊 Types of Chains

| Chain Type | Work / Purpose (Kaam) | Use Case Example |
| :--- | :--- | :--- |
| **Simple Chain** | Ek prompt ko single model se guzaar kar output dena. | Quick Q&A |
| **Sequential Chain** | Ek model ke output ko doosre model ka input banana. | Story plot se detailed script likhna |
| **Router Chain** | Input dekh kar faisla karna ke kis specialized model ya prompt ke paas bhejain. | Math questions ko math prompt, general ko general prompt pe bhejna |
| **Retrieval (RAG) Chain** | External documents (PDFs, Databases) se pehle data nikalna, phir model ko context dena. | Chat with your own PDF Book |

## 🔗 Explanations & Codes of Different Chains

Ab hum har tarah ki chain ko asaan Roman Urdu mein samjhenge aur un ke basic codes dekhenge.

### 1. Simple Chain
Yeh sabse basic chain hoti hai jo sirf aik single input leti hai, usay prompt template mein dalti hai aur model se response generate karwati hai (Jaise hum ne upar Gemini ke sath execute kiya).

```python
# Simple Chain creation
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template("Write a short poem about {topic}")


simple_chain = prompt | groq_model | StrOutputParser()
print("Simple Chain created successfully!")
```

```python
simple_chain.invoke("Ai")
```

### 2. Sequential Chain
Is chain ka kaam yeh hota hai ke pehle model/prompt ka output automatically doosri chain ka input ban jata hai. Jaise step-by-step pipeline banana.

```python
# Sequential Chain creation
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt1 = ChatPromptTemplate.from_template("Suggest a catchy book title for a topic about {topic}")
prompt2 = ChatPromptTemplate.from_template("Write a 2-sentence summary of a book titled: {book_title}")

# chain1 = prompt1 | gemini_model| StrOutputParser()
chain2 = prompt1 | groq_model | StrOutputParser()| prompt2  | groq_model | StrOutputParser()


sequential_chain = chain2.invoke({"topic": "Ai"})
```

```python
import langchain
# Dynamically set verbose attribute to fix legacy ConversationChain bug in LangChain v0.3+
langchain.verbose = False

from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationChain

# Initialize Memory
memory = ConversationBufferMemory()

# Initialize ConversationChain with Model and Memory
conversation = ConversationChain(
    llm=groq_model,
    memory=memory
)

# First interaction
response1 = conversation.predict(input="Hi, my name is Hamza.")
print("AI:", response1)
```

```python
# Second interaction (testing memory)
response2 = conversation.predict(input="Do you remember my name?")
print("AI:", response2)
```
